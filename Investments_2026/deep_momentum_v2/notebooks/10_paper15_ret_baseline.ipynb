# Paper15 baseline: RF/XGBoost, five/ten years, executable RET

This side experiment uses historical U.S. S&P500/400 members and the15agreed
paper momentum/state inputs. Each arm requires its matching5/10years of complete
past dailyOHLCand uses matching5/10year classifier training and RET payoffs.
The target is the **daily cash/share path return from the next trading open to
the following monthly open**, then ranked into ten classes. It is not the
calendar-close label. No tuning or expanded feature bank.

There are48monthly forecasts in2020–2023; the paired economic comparison is
3February2020–29December2023,47months/985sessions. January's10year payoff history
is not complete at its December signal. December2023's forward label requires
protectedJanuary2024and remains unavailable. No shortened terminal close is
substituted into that learning label. Both estimators share each arm's same
cohort, target bins and causal payoff vectors.

See [the complete frozen method](../docs/PAPER15_RET_BASELINE_20261026.md).
All cells below read saved small reports after independent proofs. **RunAll
never fits, loads a model, rebuilds data or launches a backtest.** Raw data and
private results are absent from a fresh clone. Source identities, history
completeness, adjustment vintage, action coverage and borrowing remain limits.


```python
from pathlib import Path
import json
import sys

ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "src/dmv2").is_dir())
sys.path.insert(0, str(ROOT / "src"))
local_config = ROOT / "configs/paper15_baseline.local.json"
private = json.loads(local_config.read_text()) if local_config.exists() else {}
BASELINE_RUN = Path(private.get("output_directory", ROOT / "outputs/paper15_ret_baseline/20261026-v3"))
print("Workspace:", ROOT)
print("Saved run:", BASELINE_RUN)
```
